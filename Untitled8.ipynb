```python
d = {'name': 'Ram', 'age': 20}

key = input("Enter key: ")

if key in d:
    print("Key exists")
else:
    print("Key does not exist")
```

```python
d = {'a': 50, 'b': 20, 'c': 40, 'd': 10}

asc = dict(sorted(d.items(), key=lambda x: x[1]))
desc = dict(sorted(d.items(), key=lambda x: x[1], reverse=True))

print("Ascending:", asc)
print("Descending:", desc)
```
